```python
import pandas as pd
import numpy as np
s=pd.Series([1,2,5,np.nan,None,8,9],index=['A','B','C','D','E','F','G'],name='learning')#np.nan代表缺失值，None代表原生值,NaN代表缺失值的统一处理
print(s)
s.tail()#默认取前五行
s.tail()#默认取后五行
```

```python
s.describe()#描述性信息，去掉NaN即忽略缺失值
```

```python
s.count() #忽略缺失值
#获取索引
print(s.keys())#方法
print(s.index)#属性
```

```python
s.isna()#检查缺失值
s.isin([1,2,3])#检查某个值
```

```python
print(s.sort_values())#排序
print(s.quantile(0.7))#分位数
#——————————
#1 2 5 8 9
#位置4*0.7=2.8
#值的计算 5+（8-5）*0.8=7.4
```

```python
#众数
s.mode()#可加print单独执行
s.value_counts()#众数的统计
```

```python
s.drop_duplicates()#去重
```

```python
print(s.sort_index())#按索引排序
print(s.sort_values())#按值排序
```

```python
np.random.seed(10)
score=pd.Series(np.random.randint(50,101,10),index=['学生'+str(i) for i in range(1,11)],name='成绩')
print(score)
print(score.mean())#平均值
score.max()#最大值
#高于平均分的人数
mean=score.mean()
print('高于平均分人数',len(score[score>mean]))
print('高于平均分的人数',score[score>mean].count())
```

```python
import numpy as np
import pandas as pd
t1=pd.Series([28,25,26,30,31,35,29],index=['周一','周二','周三','周四','周五','周六','周日'])
t1.sort_index()#排序，也可以使用values
d=t1.mean()
print(t1[t1>d])
```

```python
d=t1.diff().abs()#计算温度的绝对差值
print(d.sort_values(ascending=False))#绝对差值排序
print(d.index)#索引排序
print(d.keys())#索引方法2排序prt
```

```python
import numpy as np
import pandas as pd
dates=pd.date_range('2026.04.09',periods=7)
print(list(dates))
```

```python
prices=pd.Series([100,102,103,105,106,107,108],index=dates,name='prices')
print(prices)#每日价格统计
```

```python
b=prices.pct_change() #（每日利润率统计，pct代表percent）
print(b)
```

```python
b.idxmin()#最低收益率，索引
```

```python
b.mean()
```

```python
#某产品过去12个月的销量
b= pd.date_range('2026-01-01',periods=12,freq='MS')#frep代表周期，M代表月，S代表月初第一天，需大写
print(b)
sales=pd.Series([110,120,125,130,145,152,165,187,158,189,198,212],index=b)
print(sales)
```

```python
q=sales.resample('QS').mean().astype(int)#季度平均销量，Q代表季度quarter,S代表start,resample代表重新取样,astype(int)代表取整
print(q)
q1=sales.pct_change()
print(q1)
c=q1>0
print(c)
c[c.rolling(3).sum()==3].index.tolist#rolling代表滚动窗口，计算连续超过两个月增长的月份
```
